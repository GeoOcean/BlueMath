# ESTELA application

**ESTELA** (Evaluation of Source and Travel time of wave Energy reaching a Local Area) method from [Pérez et al., 2014](https://doi.org/10.1007/s10236-014-0740-7), identifies the relevant amount of energy directed along great circle arcs towards the study site using the partitioned wave spectra from CAWCR hindcast at a global scale.


- The python library used for generating the ESTELA can be found at: https://github.com/jorgeperezg/olas


```python
sites_dict = {
    "Boso": {"lon": 140.58, "lat": 34.90, "name": "Boso (Japan)"},
}
```

```python
from bluemath_tk.waves.estela import ESTELA

est = ESTELA(p_waves="https://geoocean.unican.es/thredds/catalog/rawData/CSIRO/CAWCR/CSIRO_partitions_ESTELA", source="cawcr")

est.set_site(
    site="Boso", location=(sites_dict["Boso"]["lon"] % 360, sites_dict["Boso"]["lat"])
)

est.set_years(years=[2023, 2024])
```

```python
est.PATHS
```

```python
import logging
import warnings

logging.getLogger().setLevel(logging.WARNING)
warnings.filterwarnings("ignore")

est.run(nblocks=2, groupers=["1M", "ALL", "time.month", "time.season"], only_sea=True, overwrite=True)
```

```python
est.plot_mean_estela(
    custom_params={
        "central_longitude": sites_dict["Boso"]["lon"],
        "extent": (100, 310, -70, 70),
        "figsize": [16,10]
    }
)
```

```python
est.plot_grouped_estela(
    custom_params={
        "central_longitude": sites_dict["Boso"]["lon"],
        "extent": (100, 310, -70, 70),
        "figsize": [16,10]
    },
    times=["2024-01", "2024-02", "2024-03", "2024-04", "2024-05", "2024-06", "2024-07", "2024-08", "2024-09"],
    anomaly = True,
    ncols = 3,
    max_panels = 10,
    )
```

```python
est.plot_monthly_estela(
    anomaly=False,
    custom_params={
        "central_longitude": sites_dict["Boso"]["lon"],
        "extent": (100, 310, -70, 70),
        "fontsize": 11,
    },
)
```

```python
est.plot_season_estela(
    anomaly=True,
    custom_params={
        "central_longitude": sites_dict["Boso"]["lon"],degrees
        "extent": (100, 310, -70, 70),
        "figsize": [16, 12],
    },
)
```
