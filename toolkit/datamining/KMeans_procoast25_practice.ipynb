# MDA

## 2. Seleccionar 50 casos mediante MDA de las variables ['pres', 'sst', 'mld']

```python
# subset size, scalar and directional indexes
n_subset = 50           # subset size
ix_scalar = [0, 1, 2]        # x,y,z
ix_directional = []      # 
```

```python
matrix_mda = np.vstack([data.pres.values, data.sst.values, data.mld.values]).T
matrix_mda.shape #[Number of points] x [Number of dimensions]
```

```python
# MDA algorithm
sel = MaxDiss_Simplified_NoThreshold(matrix_mda, n_subset, ix_scalar, ix_directional)
```

# KMA

## 4. Crear matriz de datos y normalizar

```python
matrix_kmeans = np.vstack([data.pres.values, data.sst.values, data.mld.values]).T
matrix_kmeans.shape #[Number of points] x [Number of dimensions]
```

```python
# normalize scalar and directional data

ix_scalar = [0,1,2]
ix_directional = []
matrix_kmeans_norm, minis, maxis = Normalize(matrix_kmeans, ix_scalar, ix_directional)
```

## 5. Hacer 50 grupos mediante KMA y desnormalizar los centroides

```python
# KMEANS

n_subset = 50
kma = KMeans(n_clusters=n_subset, n_init=100).fit(matrix_kmeans_norm)
bmus = kma.labels_
centroids_norm  = kma.cluster_centers_

centroids = DeNormalize(centroids_norm, ix_scalar, ix_directional, minis, maxis)
```

## 6. Pintar gráfico 3D de los datos y los centroides

- El color de cada punto será el cluster al que pertenece cada dato

```python
fig = plt.figure(figsize=(15, 8))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(data.pres.values, data.sst.values, data.mld.values, c = bmus, alpha=.3, cmap = 'rainbow')

ax.scatter(centroids[:,0], centroids[:,1], centroids[:,2], s=80, c = range(n_subset),
           alpha=1, cmap = 'rainbow', edgecolor = 'k')
ax.set_xlabel('Pressure', fontsize = 20)
ax.set_ylabel('SST', fontsize = 20)
ax.set_zlabel('MLD', fontsize = 20)
```

## 9. Crear una matriz de datos tridimensional homogenea de 30 datos equiespaciados entre el minimo y el mínimo y el máximo de cada variable

- Una vez creadas las matrices, transformar a vector

```python
pres_r = np.linspace(data.pres.min(), data.pres.max(), 30)
sst_r = np.linspace(data.sst.min(), data.sst.max(), 30)
mld_r = np.linspace(data.mld.min(), data.mld.max(), 30)

pres_r, sst_r, mld_r = np.meshgrid(pres_r, sst_r, mld_r)

pres_r = pres_r.reshape(-1)
sst_r = sst_r.reshape(-1)
mld_r = mld_r.reshape(-1)
```

## 10. Crear matriz de KMA y normalizar

```python
matrix_ = np.vstack([pres_r, sst_r, mld_r]).T
```

```python
matrix_norm_, _, _= Normalize(matrix_, ix_scalar, ix_directional, minis = minis, maxis = maxis)
```

## 11. Predecir a que cluster corresponde cada dato

```python
bmus_pred = kma.predict(matrix_norm_)
bmus_pred
```

```python
predicted_points = centroids[bmus_pred,:]
```
