# Optimal Threshold Selection 

Select optimal threshold for an exponential randomly generated data using the Studentized Residuals Method, Mínguez (2025) [1]

```python
data = np.random.exponential(scale=2.1, size=20*365) # 20 years of daily data
```

```python
threshold = 0 # Initial threshold
n0 = 10 # Minimum number of exceedances
min_peak_distance = 2 # 2 days between peaks
siglevel=0.05

# Inicializar la clase
opt_thres = OptimalThreshold(
    data=data,
    method="studentized",
    plot=True
)

# Obtener los picos
threshold, pks, pks_idx = opt_thres.fit()
```

```python
opt_thres.potplot()
```
